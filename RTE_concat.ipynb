# This Notebook aggregate the data from RTE's for each years and for 'en cours'

```python
# Imports for data
import numpy as np
import pandas as pd

PATH_RTE = 'data/RTE/'
regions = [
    'Auvergne-Rhône-Alpes', 'Bourgogne-Franche-Comté', 'Bretagne',
    'Centre-Val-de-Loire', 'Grand-Est', 'Hauts-de-France', 'Ile-de-France',
    'Normandie', 'Nouvelle-Aquitaine', 'Occitanie', 'PACA', 'Pays-de-la-Loire'
]

col_to_keep=['Périmètre',	'Nature',	'Date',	
             'Heures',	'Consommation',	'Thermique',	
             'Nucléaire',	'Eolien',	'Solaire',	
             'Hydraulique',	'Pompage',	'Bioénergies',	
             'Ech. physiques'
             ]



for year in range(2013, 2025):
    df_list = []
    for region in regions:
        file = f"{PATH_RTE}Regions/eCO2mix_RTE_{region}_Annuel-Definitif_{year}.xls"
        df = pd.read_csv(file, sep='\t', encoding='latin1', index_col=False, usecols=col_to_keep)
        df = df.replace('ND', np.nan)
        df['region'] = region
        df_list.append(df)
    
    yearly_df = pd.concat(df_list, ignore_index=True)
    yearly_df.to_csv(f"{PATH_RTE}rte_regions_{year}.csv", index=False)
    print(f"{year} done — {len(yearly_df)} rows")
```

```python
# Saving en cours
df_list = []
for region in regions:
    file = f"{PATH_RTE}Regions/eCO2mix_RTE_{region}_En-cours-Consolide.xls"
    df = pd.read_csv(file, sep='\t', encoding='latin1', index_col=False, usecols=col_to_keep)
    df = df.replace('ND', np.nan)
    df['region'] = region
    df_list.append(df)

    en_cours_df = pd.concat(df_list, ignore_index=True)
    en_cours_df.to_csv(f"{PATH_RTE}/rte_regions_en_cours.csv", index=False)
    print(f"{region} done — {len(region)} rows")
```
